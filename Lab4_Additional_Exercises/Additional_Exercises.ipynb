```python

# 2. STUDENT PERFORMANCE DATA ANALYZER


LAST_NAME = "CELO"
SEED_NUM = 1
FAVORITE_ARTIST = "FELIP"

# 2. Organized Student Data
labels = ["ALPHA", "BETA", "CHARLIE", "DELTA", "ECHO"]
student_labels = [f"{LAST_NAME} {word}" for word in labels]
print(f"Generated Test Data: {student_labels}")

# Values for Scores
favorite_artist = int(FAVORITE_ARTIST.index("F"))
favorite_artists = int(FAVORITE_ARTIST.index("E"))
seed_num = int(SEED_NUM + 5)

# 1. Generate Performance Records
def student_grade(seed_num, SEED_NUM):
    student_grades = []
    for index in range(len(labels)):
        a = (index % 5) + 1
        base_grade = 65 + (a * seed_num) + SEED_NUM
        b = min(100, max(50, base_grade + favorite_artists))
        c = min(100, max(50, base_grade - favorite_artist + 5))
        d = min(100, max(50, base_grade + 2))
        e = min(100, max(50, base_grade - 3))
        f = None if index == 4 else min(100, max(50, base_grade + 4))

        student_grades.append({
            "PROGRAMMING": b,
            "MATHEMATICS": c,
            "ELECTRONICS": d,
            "PHYSICS": e,
            "ENGLISH": f
        })
    return student_grades

# 2. Organized Subject Data
print(f"\n===== Organized Student/Subject Data =====")
print(f"PROGRAMMING")
print(f"MATHEMATICS")
print(f"ELECTRONICS")
print(f"PHYSICS")
print(f"ENGLISH")

# 4. Performace Categories (From Lab Activity 3)
def performance_categories(scores):
    return sum(scores) / len(scores)
def assign_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 80:
        return "B"
    elif avg >= 70:
        return "C"
    elif avg >= 60:
        return "D"
    else:
        return "F"
def generate_remark(grade):
    remarks = {
        "A": "Excellent Perfromance",
        "B": "Good Performance",
        "C": "Satisfactory Performance",
        "D": "Needs Improvement",
        "F": "Failing Status"
    }
    return remarks.get(grade, "Invalid Grade")

print(f"\n===== Performance Categories =====")
print(f"A: Excellent Perfromance")
print(f"B: Good Performance")
print(f"C: Satisfactory Performance")
print(f"D: Needs Improvement")
print(f"F: Failing Status") 

# 3. Repeated/Incomplete Record
performance_records = student_grade(seed_num, SEED_NUM)
student_labels.append(f"{LAST_NAME} ALPHA")
print(f"\nGenerated Test Data with Repeatition: {student_labels}")

print(f"\n===== Repeated/Incomplete Records =====")
students_grades = set()
fix = []
fix_students = []
fix_scores = []

incomplete = 0
repeated = 0
for index, student in enumerate(student_labels):
    data_index = index if index < 5 else 0
    scores = performance_records[data_index]

    # For Incomplete Record
    has_incomplete = any(score_val is None for score_val in scores.values())
    if has_incomplete and student not in fix:
        incomplete += 1
        fix.append({"type": "Incomplete", "student": student})
        print(f"Incomplete Record: {student}")

    # For Repeated Record
    if student in students_grades:
        repeated += 1
        fix.append({"type": "Duplicate", "student": student})
        print(f"Repeated Student-Subject Record: {student}")
        continue
    else:
        students_grades.add(student)
        fix_students.append(student)
        fix_scores.append(scores)
repeat = len([x for x in fix if x["type"] == "Repeated"])
incompletes = len([x for x in fix if x["type"] == "Incomplete"])

print(f"Incomplete Record: {incompletes}")
print(f"Repeated Record: {repeat}")

# 5. Student ans Subject Summaries
print(f"\n===== Student and Subject Summaries =====")
SUBJECTS = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

for index, student in enumerate(fix_students):
    print(f"\nStudent: {student}")
    scores = fix_scores[index]
    
    for subject in SUBJECTS:
        subjects_grade = scores[subject]
        final_grade = f"{subjects_grade}" if subjects_grade is not None else "MISSING"
        print(f"   {subject:<13} Grade: {final_grade}")

# Ordered Results
print(f"\n===== Final Student Performance Report =====")
for index, student in enumerate(fix_students):
    scores = fix_scores[index]
    subjects_grades = [val for val in scores.values() if val is not None]
    final_grades = performance_categories(subjects_grades) if subjects_grades else 0
    grade_category = assign_grade(final_grades)
    remarks = generate_remark(grade_category)
    
    print(f" {student:<15}")
    print(f"   Average Score: {final_grades:.2f}% | Grade Bracket: {grade_category} -> {remarks}")
    print("-" * 75)
```

```python
# TRANSANCTION HISTORY ANALYZER


LAST_NAME = "CELO"
SEED_NUM = 1
FAVORITE_ARTIST = "FELIP"

# 1. Generated Transanction Records
print(f"\n===== Generated Transanction Records =====")
customers = [f"{LAST_NAME} C{x}" for x in range(1, 6)]
print(f"Generated Transanction Records: {customers}")

# Generated Transanction Records with Repeatition
# 3. Repeated/Distinct Transanctions
print(f"\n===== Repeated/Distinct Transanctions =====")
customers.append("CELO C1")
print(f"Generated Transanction Records with Repeatition: {customers}")

# 2. Organized Transanction Data
print(f"\n===== Organized Transanction Data =====")
print(f"NOTEBOOK")
print(f"PEN")
print(f"USB")
print(f"HEADPHONES")
print(f"CALCULATOR")

# Prices
NOTEBOOK = 25.75
PEN = 15.25
USB = 78.35
HEADPHONES = 500.45
CALCULATOR = 2500.15

# 1. Transanction Values
favorite_artist = int(FAVORITE_ARTIST.index("F")) # Equals 0
favorite_artists = int(FAVORITE_ARTIST.index("E")) # Equals 1
seed_num = int(SEED_NUM + 9)

def product(seed_num, SEED_NUM):
    price = []
    
    for index in range(5):
        a = (index % 5) + 1
        base_qty = (((a * seed_num) + SEED_NUM) % 5) + a
        
        notebook = max(1, base_qty + favorite_artists)
        pen = max(1, base_qty - favorite_artist + 1)
        usb = max(1, base_qty + 2)
        headphones = max(1, base_qty - 1)
        calculator = max(1, base_qty + 3)

        price.append({
            "Notebook:": notebook,
            "Pen:": pen,
            "USB:": usb,
            "Headphones:": headphones,
            "Calculator:": calculator
        })
    return price

transanction_records = product(seed_num, SEED_NUM)


# 3. Repeated/Distinct Transactions
print(f"\n===== Repeated Records Auditing =====")
organized_costumer = set()
fix = []
fix_customer = []
fix_quantity = []

repeated = 0

for index, customer in enumerate(customers):
    data_index = index if index < 5 else 0
    quantities = transanction_records[data_index]

    # For Repeatition
    if customer in organized_costumer:
        repeated += 1
        fix.append({"type": "Repeated", "customer": customer})
        print(f"Repeated Student-Subject Record: {customer}")
        continue
    else:
        organized_costumer.add(customer)
        fix_customer.append(customer)
        fix_quantity.append(quantities)

repeat = len([x for x in fix if x["type"] == "Repeated"])
print(f"Repeated Transanction: {repeat}")

# 5. Costumer Summaries  
# 6. Product Summaries  
print(f"\n===== Customer and Product Summaries =====")

PRICES = {
    "Notebook:": NOTEBOOK,
    "Pen:": PEN,
    "USB:": USB,
    "Headphones:": HEADPHONES,
    "Calculator:": CALCULATOR
}

product_total = {prod: 0.0 for prod in PRICES}
product_count = {prod: 0 for prod in PRICES}
global_gross_sales = 0.0

# Reciept
for index, customer in enumerate(fix_customer):
    print(f"\nCUSTOMER: {customer}")
    print(f"{'PRODUCT':<13} {'PRICE PER UNIT':<20} {'QUANTITY':<16} {'TOTAL PRICE':<15}")
    print("-" * 68)
    
    quantities = fix_quantity[index]
    due_amount = 0.0
    
    for product in PRICES:
        quantity = quantities[product]
        unit_price = PRICES[product]
        
        total_price = quantity * unit_price
        due_amount += total_price
        
        product_total[product] += total_price
        product_count[product] += quantity
        
        print(f"{product:<13} {unit_price:<20.2f} {quantity:<16} {total_price:<15.2f}")
            
    print(f"{'DUE AMOUNT':<53} {due_amount:,.2f}")


# 6. Ordered Results and Overall Transaction Summary
print(f"\n===== Final Consolidated Transaction Report =====")
for index, customer in enumerate(fix_customer):
    quantities = fix_quantity[index]
    due_amount = sum(quantities[prod] * PRICES[prod] for prod in PRICES)
    print(f" {customer:<15} Total Due Amount: PHP {due_amount:,.2f}")

```
